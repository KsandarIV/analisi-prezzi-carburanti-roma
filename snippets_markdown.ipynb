# <span style="color:#74EE15">**2. TAG DI STATO**</span>

## <span style="color:#007ACC">**TAG DI STATO**</span>

I tag stanno a fine riga, tra parentesi quadre, e dicono a che punto è una voce. Un tag non si cancella quando la voce è chiusa: gli si affianca [<span style="color:#74EE15">**done**</span>], così resta la storia di cosa è successo.

| Tag | Significato | Quando si usa | Esempio |
|---|---|---|---|
| [<span style="color:#FF0000">**to_check**</span>] | da verificare | un dato o un'ipotesi di cui non sono ancora sicuro | <span style="color:#B026FF">**Nomentana 490**</span>: due impianti a <span style="color:#007ACC">**7 km**</span> con lo stesso indirizzo [<span style="color:#FF0000">**to_check**</span>] |
| [<span style="color:#FF007F">**to_manage**</span>] | decisione da prendere | ho trovato un problema e devo scegliere cosa farne | <span style="color:#007ACC">**2**</span> coppie con nome e indirizzo uguali [<span style="color:#FF007F">**to_manage**</span>] |
| [<span style="color:#74EE15">**done**</span>] | risolto | la verifica o la decisione è completata; si aggiunge accanto all'altro tag | <span style="color:#007ACC">**121**</span> impianti senza prezzi [<span style="color:#FF0000">**to_check**</span>] [<span style="color:#74EE15">**done**</span>] |
| [<span style="color:#FF5500">**to_do**</span>] | da fare in una fase successiva | la decisione è presa ma l'azione viene dopo | dataset dei prezzi extra da creare in pulizia [<span style="color:#FF5500">**to_do**</span>] |
| [<span style="color:#B026FF">**ML**</span>] | da usare nella parte di machine learning | il caso serve come test, previsione o controllo nella sezione ML | impianti senza prezzi: prediction set [<span style="color:#B026FF">**ML**</span>] |
| [<span style="color:#B026FF">**EDIT**</span>]{<span style="color:#B026FF">**n**</span>} | modifica postuma | cambio qualcosa che avevo già scritto più su | vedi sotto |

**Come si usa [<span style="color:#B026FF">**EDIT**</span>]{<span style="color:#B026FF">**n**</span>}:** il numero è progressivo (<span style="color:#B026FF">**001**</span>, <span style="color:#B026FF">**002**</span>, …) e collega due punti del notebook.
1. Nella cella dove la cosa era già stata scritta aggiungo, a fine riga, il rimando → causa di [<span style="color:#B026FF">**EDIT**</span>]{<span style="color:#B026FF">**002**</span>}
2. Nel punto in cui faccio la modifica metto una cella che inizia con [<span style="color:#B026FF">**EDIT**</span>]{<span style="color:#B026FF">**002**</span>}, con una riga su cosa cambia e perché.

**Esempio:** in pulizia scopro che gli impianti da gestire non sono più nel df.
